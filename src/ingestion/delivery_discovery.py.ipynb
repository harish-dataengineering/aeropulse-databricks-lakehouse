```python
"""
AeroPulse Enterprise Lakehouse Platform.

Source delivery discovery utilities.

This module discovers source delivery directories from the
Unity Catalog raw landing Volume.
"""

from typing import List


def discover_source_deliveries(
    dbutils,
    source_path: str,
    source_entity: str,
) -> List[str]:
    """
    Discover timestamped source delivery directories.

    Parameters
    ----------
    dbutils:
        Databricks utilities object.
    source_path:
        Base source landing path.
    source_entity:
        Source business entity name.

    Returns
    -------
    List[str]
        Sorted list of source delivery paths.
    """

    source_items = dbutils.fs.ls(source_path)

    delivery_prefix = f"{source_entity}_"

    delivery_paths = [
        item.path
        for item in source_items
        if item.isDir()
        and item.name.startswith(delivery_prefix)
    ]

    return sorted(delivery_paths)
```
