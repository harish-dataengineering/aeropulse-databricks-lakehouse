```python
"""
AeroPulse Enterprise Lakehouse Platform.

Synthetic engine source generator.

This module simulates engine master data received
from the ERP source system.
"""

from pyspark.sql import SparkSession, DataFrame
from pyspark.sql import functions as F


def generate_engines(
    spark: SparkSession,
    record_count: int = 2000,
) -> DataFrame:
    """
    Generate synthetic engine master data.

    Parameters
    ----------
    spark:
        Active SparkSession.
    record_count:
        Number of engine records to generate.

    Returns
    -------
    DataFrame
        Synthetic engine source data.
    """

    df = (
        spark.range(1, record_count + 1)

        .withColumn(
            "engine_id",
            F.format_string(
                "EN%09d",
                F.col("id")
            )
        )

        .withColumn(
            "engine_serial_number",
            F.concat(
                F.lit("SN-"),
                F.format_string(
                    "%010d",
                    F.col("id")
                )
            )
        )

        .withColumn(
            "aircraft_id",
            F.format_string(
                "AC%08d",
                ((F.col("id") - 1) % 1000) + 1
            )
        )

        .withColumn(
            "engine_model",
            F.when(
                F.col("id") % 3 == 0,
                F.lit("Trent XWB")
            )
            .when(
                F.col("id") % 3 == 1,
                F.lit("GE90")
            )
            .otherwise(
                F.lit("PW1100G")
            )
        )

        .withColumn(
            "engine_manufacturer",
            F.when(
                F.col("id") % 3 == 0,
                F.lit("Rolls-Royce")
            )
            .when(
                F.col("id") % 3 == 1,
                F.lit("GE Aerospace")
            )
            .otherwise(
                F.lit("Pratt & Whitney")
            )
        )

        .withColumn(
            "manufacture_year",
            (
                F.lit(2018)
                + (F.col("id") % 8)
            ).cast("integer")
        )

        .withColumn(
            "engine_status",
            F.when(
                F.col("id") % 25 == 0,
                F.lit("MAINTENANCE")
            )
            .otherwise(
                F.lit("ACTIVE")
            )
        )

        .withColumn(
            "total_operating_hours",
            (
                F.lit(1000)
                + (F.col("id") * 17)
            ).cast("long")
        )

        .withColumn(
            "total_cycles",
            (
                F.lit(500)
                + (F.col("id") * 11)
            ).cast("long")
        )

        .drop("id")
    )

    return df
```
