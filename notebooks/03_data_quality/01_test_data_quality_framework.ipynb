```python
test_df = spark.createDataFrame(
    [
        ("M001", "A001"),
        ("M002", "A002"),
        ("M003", None),
        ("M003", "A003"),
    ],
    [
        "maintenance_id",
        "aircraft_id"
    ]
)

display(test_df)
```

```python
rules = [
    DataQualityRule(
        rule_name="maintenance_id_not_null",
        description="Maintenance ID must not be NULL",
        check_function=null_check("maintenance_id")
    ),
    
    DataQualityRule(
        rule_name="aircraft_id_not_null",
        description="Aircraft ID must not be NULL",
        check_function=null_check("aircraft_id")
    ),
    
    DataQualityRule(
        rule_name="maintenance_id_unique",
        description="Maintenance ID must be unique",
        check_function=duplicate_check("maintenance_id")
    ),
]
```

```python
results = []

for rule in rules:
    result = run_data_quality_rule(
        test_df,
        rule
    )

    results.append(result)

results_df = spark.createDataFrame(results)

display(results_df)
```
