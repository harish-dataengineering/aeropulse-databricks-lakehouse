```python
"""
AeroPulse Enterprise Lakehouse Platform.

Synthetic ERP airline source-data generator.

This module simulates an external ERP source system by generating
airline reference data and writing the output as CSV files.
"""

import random
from datetime import datetime, timezone
from typing import List

from pyspark.sql import SparkSession


AIRLINES = [
    ("British Airways", "BAW", "United Kingdom", "Europe"),
    ("Lufthansa", "DLH", "Germany", "Europe"),
    ("Air France", "AFR", "France", "Europe"),
    ("KLM", "KLM", "Netherlands", "Europe"),
    ("Singapore Airlines", "SIA", "Singapore", "Asia"),
    ("Emirates", "UAE", "United Arab Emirates", "Middle East"),
    ("Qatar Airways", "QTR", "Qatar", "Middle East"),
    ("Qantas", "QFA", "Australia", "Oceania"),
    ("Air India", "AIC", "India", "Asia"),
    ("Delta Air Lines", "DAL", "United States", "North America"),
    ("United Airlines", "UAL", "United States", "North America"),
    ("American Airlines", "AAL", "United States", "North America"),
]


def generate_airlines(
    spark: SparkSession,
    record_count: int = 100,
) :
    """
    Generate synthetic airline master data.

    Parameters
    ----------
    spark:
        Active SparkSession.
    record_count:
        Number of synthetic airline records to generate.

    Returns
    -------
    DataFrame
        Spark DataFrame containing synthetic airline source records.
    """

    if record_count <= 0:
        raise ValueError("record_count must be greater than zero.")

    records: List[tuple] = []

    for sequence_number in range(1, record_count + 1):

        airline = random.choice(AIRLINES)

        airline_id = f"AL{sequence_number:06d}"

        airline_name = airline[0]
        airline_code = airline[1]
        country = airline[2]
        region = airline[3]

        founded_year = random.randint(1920, 2020)

        fleet_size = random.randint(20, 600)

        status = random.choices(
            ["ACTIVE", "INACTIVE"],
            weights=[95, 5],
        )[0]

        source_updated_timestamp = datetime.now(timezone.utc)

        records.append(
            (
                airline_id,
                airline_code,
                airline_name,
                country,
                region,
                founded_year,
                fleet_size,
                status,
                source_updated_timestamp,
            )
        )

    columns = [
        "airline_id",
        "airline_code",
        "airline_name",
        "country",
        "region",
        "founded_year",
        "fleet_size",
        "status",
        "source_updated_timestamp",
    ]

    return spark.createDataFrame(records, columns)
```
