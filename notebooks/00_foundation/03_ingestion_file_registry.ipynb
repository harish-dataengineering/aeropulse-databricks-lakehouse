# AeroPulse Ingestion File Registry

## Purpose

The ingestion file registry tracks source files processed by the AeroPulse data platform.

The registry supports:

- Rerun safety
- Idempotent ingestion
- Source file lineage
- Duplicate processing prevention

```python
ENVIRONMENT = "dev"

CATALOG = "workspace"

SCHEMA = f"aeropulse_{ENVIRONMENT}"

INGESTION_FILE_REGISTRY_TABLE = (
    f"{CATALOG}.{SCHEMA}.ingestion_file_registry"
)

print(INGESTION_FILE_REGISTRY_TABLE)
```

```python
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {INGESTION_FILE_REGISTRY_TABLE}
(
    source_file_path STRING NOT NULL,
    source_file_name STRING NOT NULL,
    source_system STRING NOT NULL,
    source_entity STRING NOT NULL,
    file_format STRING NOT NULL,

    pipeline_run_id STRING NOT NULL,
    ingestion_status STRING NOT NULL,

    records_read BIGINT,
    records_inserted BIGINT,

    ingestion_start_timestamp TIMESTAMP NOT NULL,
    ingestion_end_timestamp TIMESTAMP,

    error_message STRING,

    created_timestamp TIMESTAMP NOT NULL,
    updated_timestamp TIMESTAMP NOT NULL
)
USING DELTA
""")

print(
    f"Registry table created: "
    f"{INGESTION_FILE_REGISTRY_TABLE}"
)
```

```python
spark.sql(
    f"DESCRIBE TABLE {INGESTION_FILE_REGISTRY_TABLE}"
).show(truncate=False)
```
