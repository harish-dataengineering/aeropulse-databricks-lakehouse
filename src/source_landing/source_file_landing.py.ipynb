```python
"""
AeroPulse Enterprise Lakehouse Platform.

Reusable source file landing utilities.

This module simulates source-system file deliveries by writing
source DataFrames into the Unity Catalog raw landing Volume.
"""

from datetime import datetime, timezone

from pyspark.sql import DataFrame


def land_csv_source_file(
    dataframe: DataFrame,
    target_directory: str,
    source_entity: str,
) -> str:
    """
    Land a source DataFrame as a uniquely identified CSV delivery.

    Parameters
    ----------
    dataframe:
        Source DataFrame to land.
    target_directory:
        Target directory inside the raw landing Volume.
    source_entity:
        Source business entity name.

    Returns
    -------
    str
        Full path of the landed source delivery directory.
    """

    batch_timestamp = datetime.now(
        timezone.utc
    ).strftime("%Y%m%d_%H%M%S")

    delivery_path = (
        f"{target_directory}/"
        f"{source_entity}_{batch_timestamp}.csv"
    )

    (
        dataframe.write
        .mode("overwrite")
        .option("header", "true")
        .csv(delivery_path)
    )

    return delivery_path
```
