# Audit Framework Foundation

## Purpose

The AeroPulse platform will use centralized audit logging to track pipeline execution.

The first audit table will capture pipeline execution information and will later be integrated into reusable pipeline utilities.


```python
audit_pipeline_runs_table = f"{CATALOG}.{SCHEMA}.audit_pipeline_runs"

spark.sql(f"""
CREATE TABLE IF NOT EXISTS {audit_pipeline_runs_table}
(
    pipeline_run_id STRING NOT NULL,
    pipeline_name STRING NOT NULL,
    environment STRING NOT NULL,
    layer STRING,
    source_system STRING,
    target_table STRING,
    start_timestamp TIMESTAMP NOT NULL,
    end_timestamp TIMESTAMP,
    pipeline_status STRING NOT NULL,
    records_read BIGINT,
    records_inserted BIGINT,
    records_updated BIGINT,
    records_rejected BIGINT,
    error_message STRING,
    created_timestamp TIMESTAMP NOT NULL
)
USING DELTA
""")

print(f"Table created successfully: {audit_pipeline_runs_table}")
```

```python
spark.sql(f"DESCRIBE TABLE {audit_pipeline_runs_table}").show(truncate=False)
```
