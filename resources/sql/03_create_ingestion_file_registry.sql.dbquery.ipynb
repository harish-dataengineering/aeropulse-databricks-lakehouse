```sql
-- AeroPulse Enterprise Lakehouse Platform
-- Source file ingestion registry
-- Environment: Development

CREATE TABLE IF NOT EXISTS workspace.aeropulse_dev.ingestion_file_registry
(
    source_file_path STRING NOT NULL,
    source_file_name STRING NOT NULL,
    source_system STRING NOT NULL,
    source_entity STRING NOT NULL,
    file_format STRING NOT NULL,

    pipeline_run_id STRING NOT NULL,
    ingestion_status STRING NOT NULL,

    records_read BIGINT,
    records_inserted BIGINT,

    ingestion_start_timestamp TIMESTAMP NOT NULL,
    ingestion_end_timestamp TIMESTAMP,

    error_message STRING,

    created_timestamp TIMESTAMP NOT NULL,
    updated_timestamp TIMESTAMP NOT NULL
)
USING DELTA;
```
